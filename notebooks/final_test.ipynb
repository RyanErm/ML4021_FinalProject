```python
import pandas as pd
import numpy as np
from sklearn.svm import SVR
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import GridSearchCV
```

```python
X_train = pd.read_csv("../data/X_train_data.csv")
y_train = pd.read_csv("../data/Y_train_data.csv")
X_test = pd.read_csv("../data/X_test_data.csv")
y_test = pd.read_csv("../data/Y_test_data.csv")
```

```python
svr_improved = Pipeline([
    ("scaler", StandardScaler()),
    ("svr", SVR(kernel = 'rbf', C = 1.0, epsilon = 0.1))
])
```

```python
improved = svr_improved.fit(X_train, y_train.values.ravel())
```

```python
y_pred = svr_improved.predict(X_test)
```

```python
#score
r2 = r2_score(y_test, y_pred) # Predict on the test set
print("R2 is equal to:", r2)
#score
MSE = mean_squared_error(y_test, y_pred)
print("MSE is eqaul to: ", MSE)
#score
rmse = np.sqrt(MSE)
print("SVR RMSE:", rmse)
```
