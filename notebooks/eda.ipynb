```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv("../data/X_train_data.csv")
```

```python
df.drop(["Unnamed: 0"], axis = 1, inplace = True)
df
```

```python
df.columns
```

```python
#grab all the genre column

df_genre = df[['Action', 'Adventure', 'Cars',
       'Comedy', 'Dementia', 'Demons', 'Drama', 'Ecchi', 'Fantasy', 'Game',
       'Harem', 'Hentai', 'Historical', 'Horror', 'Josei', 'Kids', 'Magic',
       'Martial Arts', 'Mecha', 'Military', 'Music', 'Mystery', 'Parody',
       'Police', 'Psychological', 'Romance', 'Samurai', 'School', 'Sci-Fi',
       'Seinen', 'Shoujo', 'Shoujo Ai', 'Shounen', 'Shounen Ai',
       'Slice of Life', 'Space', 'Sports', 'Super Power', 'Supernatural',
       'Thriller', 'Unknown', 'Vampire', 'Yaoi', 'Yuri']]
```

```python
df_genre_counts = df_genre.sum().sort_values(ascending=False).reset_index()

labels = df_genre_counts.head(10)
```

The top ten genres represented in this dataset

```python
plt.bar(x=labels['index'], height = labels[0])
plt.xticks(rotation=90)
plt.xlabel("Anime Genre")
plt.ylabel("Count of Occurences")
plt.title("Amount of occurrences for the top 10 genres")
plt.show()
```

```python
df_genre_counts = df_genre.sum().sort_values().reset_index()
min_labels = df_genre_counts.head(10)
```

```python
plt.bar(x=min_labels['index'], height = min_labels[0])
plt.xticks(rotation=90)
plt.xlabel("Anime Genre")
plt.ylabel("Count of Occurences")
plt.title("Amount of occurrences for the lowest 10 genres")
plt.show()
```

```python
year = df.groupby("Year", as_index=False).count()
counts_by_year = year.iloc[:, :2]
counts_by_year
```

```python
plt.plot(counts_by_year["Year"], counts_by_year["Episodes"])
plt.xlabel("Year")
plt.ylabel("Count")
plt.title("Counts of Anime releases by Year")
plt.show()
```
