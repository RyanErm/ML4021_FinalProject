```python
#import packages
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("anime.csv")
score = df["Score"]
X = df.drop("Score", axis=1)
col_drop = ["MAL_ID","English name","Japanese name","Premiered","Score-10","Score-9","Score-8","Score-7","Score-6","Score-5","Score-4","Score-3","Score-2","Score-1"]
X.columns
X.drop(col_drop, axis = 1, inplace = True)
X.columns
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, score, test_size=0.2, random_state=42)
```

```python
X_train.to_csv("X_train_data.csv")
```

```python
y_train.to_csv("Y_train_data.csv")
```

```python
X_test.to_csv("X_test_data.csv")
```

```python
y_test.to_csv("Y_test_data.csv")
```
